## Ingest data from landing to bronze

```python
%py
from pyspark.sql import functions as f
from pyspark.sql import types as t
```

```python
%py
# Crear widgets de parametros
dbutils.widgets.text("path_name", "")
```

```python
%py
# Obtener parametros
path_name = dbutils.widgets.get("path_name")
```

```python
%py
# ingest categoria
df_raw_categoria = spark.read.format("parquet").option("header", True).load(f"abfss://lakehouse@dlurbanstorage.dfs.core.windows.net/landing/{path_name}/")
```

```python
%py
# Añadir columna ingest_date
df_raw_categoria = df_raw_categoria.withColumn("ingest_date", f.current_timestamp())
```

```python
%py
# Crear archivo delta
df_raw_categoria.write.mode("append").format("delta").saveAsTable(f"urban_catalog.bronze.{path_name}")
```
