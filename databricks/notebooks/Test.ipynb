```python
%py
server = dbutils.secrets.get(scope = "kv-urban", key = "azsql-server")
database = dbutils.secrets.get(scope = "kv-urban", key = "azsql-database")
username = dbutils.secrets.get(scope = "kv-urban", key = "azsql-user")
password = dbutils.secrets.get(scope = "kv-urban", key = "azsql-password")
port = dbutils.secrets.get(scope = "kv-urban", key = "azsql-port")
```

```python
%py
jdbc_url = "jdbc:sqlserver://{0}:{1};database={2};user={3};password={4};trustServerCertificate=true;encrypt=optional;".format(server, port, database, username, password)
```

```python
%py
df_raw = spark.read.jdbc(jdbc_url, "ventas.VentaDetalle")
```
