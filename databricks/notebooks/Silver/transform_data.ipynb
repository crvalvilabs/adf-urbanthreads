## Transformations

```python
%py
df_dim_categoria = spark.read.table("urban_catalog.silver.categoria_raw")
```

```python
%py
df_dim_producto = spark.read.table("urban_catalog.silver.producto_raw")
```

```python
%py
df_fact_venta_encabezado = spark.read.table("urban_catalog.silver.ventaEncabezado_raw")
```

```python
%py
df_fact_venta_detalles = spark.read.table("urban_catalog.silver.ventaDetalle_raw")
```

### Create clean tables

```python
%py
# Rename columns
df_producto_renamed = df_dim_producto.withColumnRenamed("Nombre", "NombreProducto")
```

```python
%py
# select columns required
df_producto_clean = df_producto_renamed.select("ProductoID", "NombreProducto", "CategoriaID", "Genero", "Talla", "Color", "PrecioUnitario")
```

```python
%py
# Rename columns
df_categoria_renamed = df_dim_categoria.withColumnRenamed("Nombre", "NombreCategoria")
```

```python
%py
df_clean_categoria = df_categoria_renamed.select("CategoriaID", "NombreCategoria", "GeneroObjetivo")
```

```python
%py
# select columns required
df_ventas_encabezado_clean = df_fact_venta_encabezado.select("VentaID", "FechaVenta", "Canal", "TotalVenta")
```

```python
%py
# select columns required
df_ventas_detalle_clean = df_fact_venta_detalles.select("VentaDetalleID", "VentaID", "ProductoID", "Cantidad", "PrecioUnitario", "Descuento", "Subtotal")
```

### Write tables

```python
%py
df_producto_clean.write.mode("overwrite").format("delta").saveAsTable("urban_catalog.silver.producto_clean")
```

```python
%py
df_clean_categoria.write.mode("overwrite").format("delta").saveAsTable("urban_catalog.silver.categoria_clean")
```

```python
%py
df_ventas_encabezado_clean.write.mode("overwrite").format("delta").saveAsTable("urban_catalog.silver.ventas_encabezado_clean")
```

```python
%py
df_ventas_detalle_clean.write.mode("overwrite").format("delta").saveAsTable("urban_catalog.silver.ventas_detalle_clean")
```
